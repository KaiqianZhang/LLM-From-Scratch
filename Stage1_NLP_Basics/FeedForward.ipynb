### Implementation

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

import math

from dataclasses import dataclass
```

```python
@dataclass
class Args:
    D: int # embedding dimension
    hidden_D: int # hidden dimension
    dropout: float
```

```python
class FFN(nn.Module):
    """
    
    """
    def __init__(self, args):
        super().__init__()
        self.w1 = nn.Linear(args.D, args.hidden_D, bias=False)
        self.w2 = nn.Linear(args.hidden_D, args.D, bias=False)
        self.dropout = nn.Dropout(args.dropout)

    def forward(self, x):
        return self.dropout(self.w2(F.relu(self.w1(x))))
```

### Test

```python
# hyperparameters
D = 512
hidden_D = D * 4
dropout = 0.1
B = 8
T = 128

# instantiate args
args = Args(D, hidden_D, dropout)
print("dimension: ", args.D, "hidden dimension: ", args.hidden_D, "dropout: ", args.dropout)
```

```python
# input
x = torch.randn(B, T, D)

# check ffn
ffn = FFN(args)
output = ffn(x)

print("x output shape: ", output.shape)
print(x)
```
